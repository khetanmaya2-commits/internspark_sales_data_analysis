```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv('/content/superstore_final_dataset (1).csv',encoding='latin1')
```

```python
data.head(10)
```

```python
data.info()
```

```python
data.columns
```

```python
data.drop_duplicates(inplace=True)
```

```python
data.isnull().sum()
```

```python
data.fillna(0,inplace=True)
```

```python
data['Order_Date']=pd.to_datetime(data['Order_Date'],errors="coerce",dayfirst=True)
```

```python
data['Ship_Date'] = pd.to_datetime(data['Ship_Date'], errors='coerce', dayfirst=True)
```

```python
data.dtypes
```

```python
total_revenue=data['Sales'].sum()
print(total_revenue)
```

```python
average_order_value=data['Sales'].sum()/data['Order_ID'].nunique()
print(average_order_value)
```

```python
total_orders=data['Order_ID'].nunique()
print(total_orders)
```

```python
regional_sales=data.groupby('Region')['Sales'].sum()
print(regional_sales)
```

```python
country_sales=data.groupby('Country')['Sales'].sum()
print(country_sales)
```

```python
category_sales=data.groupby('Category')['Sales'].sum()
print(category_sales)
```

```python
top_products=data.groupby('Product_Name')['Sales'].sum().sort_values().head(10)
print(top_products)
```

```python
regional_sales.plot(kind='line',figsize=(10,5),marker='o')
plt.title('Region-wise Sales')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
country_sales.plot(kind='bar')
plt.title('Country-wise Sales')
plt.xlabel('Country')
plt.ylabel('Sales')

plt.show()
```

```python
category_sales.plot(kind='bar',figsize=(8,5))
plt.title("Category-wise Sales")
plt.xlabel('Category')
plt.ylabel('Sales')
plt.show()
```

```python
sub_category_sales=data.groupby('Sub_Category')['Sales'].sum()
print(sub_category_sales)
```

```python
sub_category_sales.plot(kind='bar')
plt.title("Sub-Category Sales")
plt.xlabel('Sub_Category')
plt.ylabel('Sales')
plt.show()
```

```python
top_products.plot(kind='barh')
plt.title("Top 10 Products")
plt.xlabel('Sales')
plt.ylabel('Product_Name')
plt.show()
```

```python
data['Sales'].plot(kind='hist',bins=20,figsize=(8,5))
plt.title('Sales Distribution')
plt.xlabel('Sales')
plt.show()
```

```python
data['Month'] = data['Order_Date'].dt.month
data['Month_Name'] = data['Order_Date'].dt.month_name()
data['Year'] = data['Order_Date'].dt.year
```

```python
monthly_sales=data.groupby('Month')['Sales'].sum()
print(monthly_sales)
```

```python
monthly_sales.plot(kind='line',figsize=(10,5),marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

```python
yearly_sales = data.groupby('Year')['Sales'].sum()

yearly_sales.plot(kind='bar', figsize=(8,5))

plt.title('Yearly Sales')
plt.xlabel('Year')
plt.ylabel('Sales')

plt.show()
```

```python
monthly_orders = data.groupby('Month')['Order_ID'].count()

monthly_orders.plot(kind='bar', figsize=(10,5))

plt.title('Monthly Orders')
plt.xlabel('Month')
plt.ylabel('Number of Orders')

plt.show()
```

```python
data.to_csv('cleaned_supersales_data.csv', index=False)
```
